```python
import matplotlib.pyplot as plt

def plot(dataset, name='mnist1_samples', figsize=(12, 6)):
    fig, axes = plt.subplots(2, 2, figsize=figsize)

    axes = axes.flatten()

    for i in range(4):
        axes[i].imshow(dataset[i][0], cmap='gray')
        axes[i].axis('off')

    plt.tight_layout(pad=1.0)

    plt.savefig(f'../assets/datasets/{name}.pdf', dpi=300, bbox_inches='tight')

    plt.show()
```

# ImageNet

```python
import torch
import matplotlib.pyplot as plt
import io
import math
import random
from PIL import Image
```

```python
import glob
import pandas as pd
import io


files = sorted(glob.glob("/home/shared/data/imagenet/validation-*.parquet"))

df = pd.read_parquet(files)
```

```python
def show_imgnet(idx, df=df, title=None):
    img_bytes = df.loc[idx, 'image']['bytes']
    img = Image.open(io.BytesIO(img_bytes))

    plt.imshow(img)
    plt.axis("off")
    if title != None:
        plt.title(title)

def show_img(img, title=None):
    img = img.squeeze()

    plt.imshow(img, cmap='gray')
    if title != None:
        plt.title(title)
    plt.axis('off')
```

```python
def get_imgnet_img(df, idx, size=(224, 224)): 
    img_bytes = df.loc[idx, 'image']['bytes']
    img = Image.open(io.BytesIO(img_bytes))
    img = img.resize(size, Image.BILINEAR)
    return img
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import io

def plot_imagenet(df=df, name='imagenet_samples', indices=[691, 10, 90, 120], target_class=None):
    fig, axes = plt.subplots(2, 2, figsize=(8, 8))
    axes = axes.flatten()

    if target_class != None:
        df = df[df['label'] == target_class].reset_index(drop=True)
    for i, idx in enumerate(indices):
        img = get_imgnet_img(df, idx, size=(224, 224))
        axes[i].imshow(img)
        axes[i].axis('off')

    plt.tight_layout(pad=1.0)

    plt.savefig(f'../assets/datasets/{name}.pdf', dpi=300, bbox_inches='tight')

    plt.show()
```

```python
plot_imagenet()
```

```python
plot_imagenet(name='imagenet_complex_samples', indices=[0, 20, 5, 14], target_class=91)
```
